```python
import pandas as pd
df = pd.read_csv('/content/superstore_final_dataset_(1)[1].csv', encoding='latin1')
df.head()
```

```python
# Dataset information
df.info()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Fill missing postal codes
df['Postal_Code'].fillna(df['Postal_Code'].mode()[0], inplace=True)
# Verify again
df.isnull().sum()
```

```python
# Total Revenue
total_revenue = df['Sales'].sum()
# Average Sales
average_sales = df['Sales'].mean()
# Maximum Sale
max_sale = df['Sales'].max()
# Minimum Sale
min_sale = df['Sales'].min()
print("Total Revenue:", total_revenue)
print("Average Sales:", average_sales)
print("Maximum Sale:", max_sale)
print("Minimum Sale:", min_sale)
```

```python
import matplotlib.pyplot as plt
# Sales by Category
category_sales = df.groupby('Category')['Sales'].sum()
# Bar chart
category_sales.plot(kind='bar')
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

```python
# Sales by Region
region_sales = df.groupby('Region')['Sales'].sum()
# Pie chart
region_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales by Region')
plt.ylabel('')
plt.show()
```

```python
# Top 10 products by sales
top_products = df.groupby('Product_Name')['Sales'].sum().sort_values(ascending=False).head(10)
# Horizontal bar chart
top_products.plot(kind='barh')
plt.title('Top 10 Selling Products')
plt.xlabel('Sales')
plt.ylabel('Products')
plt.show()
```

Insights

1. Technology and Furniture categories generated the highest sales.

2. West region showed the best sales performance.

3. Canon imageCLASS 2200 Advanced Copier was the top selling product.

4. Only Postal_Code column had missing values, which were cleaned successfully.

5. Visualizations helped identify sales trends and regional performance.
