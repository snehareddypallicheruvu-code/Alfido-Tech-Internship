```python
import pandas as pd
# Load dataset
df = pd.read_csv('traffic.csv')
# Show first 5 rows
df.head()
```

```python
# Dataset information
print(df.info())
# Missing values
print("\nMissing Values:\n")
print(df.isnull().sum())
```

```python
# Total sessions
total_sessions = df['linkid'].nunique()
# Total users (using country + city approximation)
total_users = df[['country', 'city']].drop_duplicates().shape[0]
# Bounce rate approximation
bounce_rate = (df['event'] == 'click').mean() * 100
# Average session duration (dummy approximation)
avg_session_duration = len(df) / total_sessions
print("Total Sessions:", total_sessions)
print("Total Users:", total_users)
print("Bounce Rate:", round(bounce_rate, 2), "%")
print("Average Session Duration:", round(avg_session_duration, 2))
```

```python
import matplotlib.pyplot as plt
# Top 10 countries
top_countries = df['country'].value_counts().head(10)
# Plot
top_countries.plot(kind='bar', figsize=(10,5))
plt.title('Top 10 Countries by Traffic')
plt.xlabel('Country')
plt.ylabel('Visits')
plt.show()
```

```python
# Top 10 cities
top_cities = df['city'].value_counts().head(10)
# Plot
top_cities.plot(kind='bar', figsize=(10,5))
plt.title('Top 10 Cities by Traffic')
plt.xlabel('City')
plt.ylabel('Visits')
plt.show()
```

```python
# Event distribution
event_counts = df['event'].value_counts()
# Pie chart
event_counts.plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title('User Event Distribution')
plt.ylabel('')
plt.show()
```

```python
# Top 10 tracks/pages
top_tracks = df['track'].value_counts().head(10)
# Horizontal bar chart
top_tracks.plot(kind='barh', figsize=(10,6))
plt.title('Top 10 Tracks / Landing Pages')
plt.xlabel('Visits')
plt.ylabel('Tracks')
plt.show()
```

```python
# Traffic by date
traffic_by_date = df['date'].value_counts().sort_index()
# Line chart
traffic_by_date.plot(figsize=(10,5))
plt.title('Website Traffic Over Time')
plt.xlabel('Date')
plt.ylabel('Visits')
plt.show()
```

Insights

1. Website traffic came from multiple countries and cities, indicating broad user engagement.
